```python
#Plot 1: Count Plot — Number of Flowers by Species
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='species')

plt.title('Count of Each Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')

plt.show()
```

```python
#Plot 2: Histogram — Distribution of Sepal Length
plt.figure(figsize=(8, 5))

plt.hist(df['sepal_length'], bins=15, edgecolor='black')

plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')

plt.show()
```

```python
#Plot 3: Histograms of All Numerical Features
df[['sepal_length']].hist(
    figsize=(12, 8),
    bins=15,
    edgecolor='black'
)

plt.suptitle('Distribution of Iris Features')
plt.tight_layout()
plt.show()
```

```python
df[['sepal_width']].hist(
    figsize=(12, 8),
    bins=15,
    edgecolor='black'
)

plt.suptitle('Distribution of Iris Features')
plt.tight_layout()
plt.show()
```

```python
df[['petal_length']].hist(
    figsize=(12, 8),
    bins=15,
    edgecolor='black'
)

plt.suptitle('Distribution of Iris Features')
plt.tight_layout()
plt.show()
```

```python
df[['petal_width']].hist(
    figsize=(12, 8),
    bins=15,
    edgecolor='black'
)

plt.suptitle('Distribution of Iris Features')
plt.tight_layout()
plt.show()
```

```python
#Plot 4: Box Plot — Detect Potential Outliers
plt.figure(figsize=(10, 6))

sns.boxplot(data=df[
    ['sepal_length', 'sepal_width',
     'petal_length', 'petal_width']
])

plt.title('Box Plot of Iris Features')
plt.ylabel('Measurement (cm)')

plt.show()
```

```python
#Plot 5: Scatter Plot — Sepal Length vs Sepal Width
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width',
    hue='species',
    s=80
)

plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')

plt.show()
```

```python
#Plot 6: Scatter Plot — Petal Length vs Petal Width
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species',
    s=80
)

plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```

```python
#Plot 7: Violin Plot — Petal Length by Species
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='species',
    y='petal_length'
)

plt.title('Petal Length Distribution by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')

plt.show()
```

```python
#Plot 8: Bar Plot — Average Petal Length by Species
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='species',
    y='petal_length',
    errorbar=None
)

plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length (cm)')

plt.show()
```

```python
#Plot 9: KDE Plot — Sepal Length Distribution by Species
plt.figure(figsize=(9, 5))

sns.kdeplot(
    data=df,
    x='sepal_length',
    hue='species',
    fill=True,
    common_norm=False
)

plt.title('Sepal Length Distribution by Species')
plt.xlabel('Sepal Length (cm)')

plt.show()
```

```python
#correlation Heatmap
# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation matrix
corr_matrix = numeric_df.corr()

# Plot correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap of Iris Dataset')
plt.show()
```
