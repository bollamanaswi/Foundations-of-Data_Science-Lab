```python
# Check for null values in each column and sum them up
df.isnull().sum()
```

```python
# Remove rows where all values are NaN
df.dropna(how='all',inplace=True)
df.isnull().sum()
```

```python
# Remove rows with null values specifically in the 'Age' column
df.dropna(subset=['Age'],inplace=True)
df.isnull().sum()
```

```python
# Remove rows from the 'Cabin' column and 'Embarked' column  that both have less than 1 non-null value
df.dropna(subset=['Cabin','Embarked'], thresh=1, inplace=True)
df.isnull().sum()
```

```python
# Remove rows that contain any null values (axis=0 is default)
df.dropna(axis=0,inplace=True)
df.isnull().sum()
```

```python
# Remove columns that contain any null values (axis=1)
df.dropna(axis=1,inplace=True)
df.isnull().sum()
```

```python
# Remove rows that contain any null values
df.dropna(how='any',inplace=True)
df.isnull().sum()
```
