```python
import pandas as pd
import numpy as np

# Create the first DataFrame
df1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Anu', 'Ravi', 'Manu', 'Priya'],
    'Marks': [85, 90, 78, 92],
    'Department': ['CSE', 'ECE', 'CSE', 'IT']
})

# Create the second DataFrame
df2 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 105],
    'Age': [19, 20, 21, 19],
    'City': ['Vizag', 'Hyderabad', 'Vijayawada', 'Guntur']
})

print("DataFrame 1:")
display(df1)

print("DataFrame 2:")
display(df2)
```

```python
#Hierarchical Indexing (MultiIndex)

#Hierarchical indexing allows you to have multiple index levels in a DataFrame.
data = {
    'Marks': [85, 90, 78, 92],
    'Attendance': [90, 95, 85, 98]
}

index = [
    ['CSE', 'CSE', 'ECE', 'ECE'],
    ['Anu', 'Ravi', 'Manu', 'Priya']
]

df = pd.DataFrame(
    data,
    index=index
)

df.index.names = ['Department', 'Student']

print(df)
```

```python
#Access Data Using Hierarchical Indexing
# Access all students in CSE
print(df.loc['CSE'])

# Access the marks of Anu
print(df.loc[('CSE', 'Anu'), 'Marks'])
```

```python
#Swap Index Levels
print(df.swaplevel())
```

```python
#Sort the Hierarchical Index
print(df.sort_index())
```

```python
#Vertical Concatenation

#Adds rows one below another.

df1 = pd.DataFrame({
    'Name': ['Anu', 'Ravi'],
    'Marks': [85, 90]
})

df2 = pd.DataFrame({
    'Name': ['Manu', 'Priya'],
    'Marks': [78, 92]
})

result = pd.concat([df1, df2], axis=0, ignore_index=True)

print(result)
```

```python
#Horizontal Concatenation

#Adds columns side by side.

df1 = pd.DataFrame({
    'Name': ['Anu', 'Ravi', 'Manu']
})

df2 = pd.DataFrame({
    'Marks': [85, 90, 78]
})

result = pd.concat([df1, df2], axis=1)

print(result)
```

```python
#Merging DataFrames

#The merge() function combines DataFrames using a common column.

students = pd.DataFrame({
    'Student_ID': [101, 102, 103],
    'Name': ['Anu', 'Ravi', 'Manu']
})

marks = pd.DataFrame({
    'Student_ID': [101, 102, 104],
    'Marks': [85, 90, 95]
})

result = pd.merge(students, marks, on='Student_ID', how='inner')

print(result)
```

```python
# Inner merge
print(pd.merge(students, marks, on='Student_ID', how='inner'))

# Left merge
print(pd.merge(students, marks, on='Student_ID', how='left'))

# Right merge
print(pd.merge(students, marks, on='Student_ID', how='right'))

# Outer merge
print(pd.merge(students, marks, on='Student_ID', how='outer'))
```

```python
#Reshaping DataFrames

#The melt() function converts data from wide format to long format.

df = pd.DataFrame({
    'Student': ['Anu', 'Ravi', 'Manu'],
    'Python': [85, 90, 78],
    'Java': [80, 88, 82]
})

print("Original DataFrame:")
display(df)

result = pd.melt(
    df,
    id_vars=['Student'],
    var_name='Subject',
    value_name='Marks'
)

print("After Melt:")
display(result)
```

```python
#The stack() method moves columns into the inner row index, producing a Series in this example.

df = pd.DataFrame({
    'Python': [85, 90],
    'Java': [80, 88]
}, index=['Anu', 'Ravi'])

result = df.stack()

print(result)
```

```python
#The unstack() method moves an index level into columns.

result = df.stack().unstack()

print(result)
```

```python
#The pivot() function rearranges data by turning unique values from one column into separate columns.

df = pd.DataFrame({
    'Student': ['Anu', 'Anu', 'Ravi', 'Ravi'],
    'Subject': ['Python', 'Java', 'Python', 'Java'],
    'Marks': [85, 80, 90, 88]
})

print("Original DataFrame:")
display(df)

result = df.pivot(
    index='Student',
    columns='Subject',
    values='Marks'
)

print("After Pivot:")
display(result)
```

```python
#Pivot Table Using pivot_table()

#The pivot_table() function summarizes data and can aggregate duplicate combinations.

df = pd.DataFrame({
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'CSE'],
    'Subject': ['Python', 'Python', 'Java', 'Java', 'Java'],
    'Marks': [80, 90, 75, 85, 95]
})

result = df.pivot_table(
    index='Department',
    columns='Subject',
    values='Marks',
    aggfunc='mean'
)

print(result)
```

```python
#Crosstabulation Using crosstab()

#The pd.crosstab() function calculates frequencies for combinations of categorical variables.

df = pd.DataFrame({
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'CSE'],
    'Gender': ['F', 'M', 'F', 'M', 'F']
})

result = pd.crosstab(df['Department'], df['Gender'])

print(result)
```

```python
#Rename Columns

#The rename() method changes column names or index labels.

df = pd.DataFrame({
    'Student_Name': ['Anu', 'Ravi'],
    'Student_Marks': [85, 90]
})

result = df.rename(columns={
    'Student_Name': 'Name',
    'Student_Marks': 'Marks'
})

print(result)
```

```python
#Sorting DataFrames

#The sort_values() method sorts rows using column values.

df = pd.DataFrame({
    'Name': ['Anu', 'Ravi', 'Manu'],
    'Marks': [85, 90, 78]
})

# Sort in ascending order
print(df.sort_values(by='Marks'))

# Sort in descending order
print(df.sort_values(by='Marks', ascending=False))
```

```python
#Dropping Duplicate Rows

#The drop_duplicates() method removes duplicate records.

df = pd.DataFrame({
    'Name': ['Anu', 'Ravi', 'Anu', 'Manu'],
    'Marks': [85, 90, 85, 78]
})

result = df.drop_duplicates()

print(result)
```

```python
# Demonstration of all types of joins in pandas
import pandas as pd

# Create sample DataFrames
df1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Anu', 'Ravi', 'Manu', 'Priya']
})

df2 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 105],
    'City': ['Vizag', 'Hyderabad', 'Vijayawada', 'Guntur']
})

print("DataFrame 1:")
display(df1)

print("\nDataFrame 2:")
display(df2)

# 1. Inner Join (Returns only matching rows in both DataFrames)
inner_join = pd.merge(df1, df2, on='Student_ID', how='inner')
print("\nInner Join:")
display(inner_join)

# 2. Left Join (Returns all rows from the left DataFrame, and matched rows from the right)
left_join = pd.merge(df1, df2, on='Student_ID', how='left')
print("\nLeft Join:")
display(left_join)

# 3. Right Join (Returns all rows from the right DataFrame, and matched rows from the left)
right_join = pd.merge(df1, df2, on='Student_ID', how='right')
print("\nRight Join:")
display(right_join)

# 4. Outer Join (Returns all rows when there is a match in either left or right DataFrame)
outer_join = pd.merge(df1, df2, on='Student_ID', how='outer')
print("\nOuter Join:")
display(outer_join)
```
