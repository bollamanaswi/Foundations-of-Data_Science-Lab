```python
import pandas as pd

# Create first DataFrame
df1 = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Name': ['Abhi', 'Sunny', 'Rishi', 'Gopi'],
    'Department': ['CSE', 'ECE', 'CSE', 'EEE']
})

# Create second DataFrame
df2 = pd.DataFrame({
    'ID': [101, 102, 103, 105],
    'Marks': [85, 90, 78, 88],
    'Grade': ['A', 'A+', 'B', 'A']
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
```

```python
# Inner Merge
print("\nInner Merge:")
print(pd.merge(df1, df2, on='ID', how='inner'))

# Left Merge
print("\nLeft Merge:")
print(pd.merge(df1, df2, on='ID', how='left'))

# Right Merge
print("\nRight Merge:")
print(pd.merge(df1, df2, on='ID', how='right'))

# Outer Merge
print("\nOuter Merge:")
print(pd.merge(df1, df2, on='ID', how='outer'))
```

```python
# Create additional DataFrames
df3 = pd.DataFrame({
    'ID': [106, 107],
    'Name': ['Eve', 'Frank'],
    'Department': ['CSE', 'ECE']
})

df4 = pd.DataFrame({
    'ID': [108, 109],
    'Name': ['Grace', 'Helen'],
    'Department': ['EEE', 'CSE']
})
```

```python
# Row-wise Concatenation
print("\nRow-wise Concatenation:")
print(pd.concat([df3, df4], ignore_index=True))

# Column-wise Concatenation
df5 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie']
})

df6 = pd.DataFrame({
    'Marks': [85, 90, 78]
})

print("\nColumn-wise Concatenation:")
print(pd.concat([df5, df6], axis=1))
```

```python
left = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie']
}, index=[101, 102, 103])

right = pd.DataFrame({
    'Marks': [85, 90, 78]
}, index=[101, 102, 104])
```

```python
result = left.join(right, how='left')

print(result)
```

```python
result = left.join(right, how='inner')

print(result)
```

```python
result = left.join(right, how='outer')

print(result)
```
