```python
df = df.drop_duplicates()
```

```python
iris_numeric = df.select_dtypes(include=['float64', 'int64'])

# Correlation matrix
corr_matrix = iris_numeric.corr()

print("Correlation Matrix:")
print(corr_matrix)
```

```python
# Heatmap
plt.figure(figsize=(6,5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of Iris Dataset")
plt.show()
```

```python
# Dispersion (Descriptive Statistics)

print("Descriptive Statistics:")
print(iris_numeric.describe())

print("\nVariance:")
print(iris_numeric.var())

print("\nStandard Deviation:")
print(iris_numeric.std())
```

```python

# Coefficient of Variation (CV)
cv = (iris_numeric.std() / iris_numeric.mean()) * 100

print("Coefficient of Variation (CV):")
print(cv)
```

```python
# Coefficient of Range

maximum = iris_numeric.max()
minimum = iris_numeric.min()

coefficient_of_range = (maximum - minimum) / (maximum + minimum)

print("Coefficient of Range:")
print(coefficient_of_range)
```

```python
# Coefficient of Quartile Deviation

Q1 = iris_numeric.quantile(0.25)
Q3 = iris_numeric.quantile(0.75)

coefficient_of_quartile_deviation = (Q3 - Q1) / (Q3 + Q1)

print("Coefficient of Quartile Deviation:")
print(coefficient_of_quartile_deviation)
```

```python
# Coefficient of Mean Deviation

mean = iris_numeric.mean()

mean_deviation = (iris_numeric - mean).abs().mean()

coefficient_of_mean_deviation = mean_deviation / mean

print("Coefficient of Mean Deviation:")
print(coefficient_of_mean_deviation)
```

```python
# Coefficient of Variation

cv = (iris_numeric.std() / iris_numeric.mean()) * 100

print("Coefficient of Variation:")
print(cv)
```
